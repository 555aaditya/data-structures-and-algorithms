<h2 style="color:#C83F49"> Classes </h2>

```python
class Cars:

    def Print(self,brand,model,engine):
        print(brand)
        print(model)
        print(engine)

car1 = Cars()  #creating Cars object named car1
car1.Print("BMW","M8","V8")  #calling a class function
```

<h2 style="color:#C83F49"> Objects</h2>

```python
class EmployeeData:
    name = "Aaditya"
    age = 20
    salary = 4800000

    def Print(self):
        print(self.name)
        print(self.age)
        print(self.salary)

#creating an object

emp1 = EmployeeData()
emp1.Print()
```

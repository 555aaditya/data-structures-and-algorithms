<h2 style="color:#C83F49"> Constructors </h2>

```python
class cls1:

    #defining the constructor
    def __init__(self):
        self.name = input("What is your name :")
        self.age = int(input("Enter your age :"))
        self.salary = int(input("What is your salary :"))

        #this calls the method when an object is created
        self.return1()

    def return1(self):
        print("\nYour name is",self.name)
        print("Your age is",self.age)
        print("You earn",self.salary)

#defining an object for the class
cls2 = cls1()
```

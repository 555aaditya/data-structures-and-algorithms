<h3 style="color:#C1A1BD">1. Basic Method</h3>

```python
class Fruit:
    def __init__(self,name):
        self._name = name

    def get_name(self):
        print("Getting name")
        return self._name

    def set_name(self,name1):
        self._name = name1


if __name__ == "__main__":
    obj1 = Fruit('Banana')
    print(obj1.get_name())
    obj1.set_name('Kiwi')
    print(obj1.get_name())
        
        
        
```

<h3 style="color:#C1A1BD">2. Decorator Method </h3>

```python
class Car:
    def __init__(self,brand):
        self._brand = brand

    @property
    def car_1(self):
        print("Getting brand")
        return self._brand

    @car_1.setter
    def car_1(self,new_brand):
        print(f"{self._brand} is now {new_brand}")
        self._brand = new_brand

    @car_1.deleter
    def car_1(self):
        print(f"{self._brand} was deleted")
        del self._brand
    


if __name__ == "__main__":
    obj1 = Car('BMW')
    print(obj1.car_1)

    obj1.car_1 = "Mercedes"
    print(obj1.car_1)

    del obj1.car_1
```

<h2 style="color:#C83F49"> Encapsulation </h2>

**Protected**
Attributes and methods prefixed with a single underscore (_) are considered protected

**Private**
Attributes and methods prefixed with double underscores (__) are considered private


<h3 style="color:#C1A1BD">1. Initialization</h3>

```python
class MyClass:
    def __init__(self):
        self.public_attribute = "I'm a public attribute"
        self._protected_attribute = "I'm a protected attribute"
        self.__private_attribute = "I'm a private attribute"

    def get_private_attribute(self):
        return self.__private_attribute

    def set_private_attribute(self, value):
        self.__private_attribute = value


# Creating an instance of MyClass
obj = MyClass()

# Accessing public, protected, and private attributes directly
print(obj.public_attribute)  # Output: I'm a public attribute
print(obj._protected_attribute)  # Output: I'm a protected attribute
# print(obj.__private_attribute)  # Raises AttributeError

# Accessing private attribute using getter method
print(obj.get_private_attribute())  # Output: I'm a private attribute

# Setting private attribute using setter method
obj.set_private_attribute("Updated private attribute")
print(obj.get_private_attribute())  # Output: Updated private attribute
```

***Public: Attributes and methods are accessible from outside the class without any restrictions.***

***Protected: Attributes and methods are intended to be accessed only within the class or its subclasses.***

***Private: Attributes and methods are intended to be accessed only within the class and cannot be accessed from outside the class or its subclasses.***


<h3 style="color:#C1A1BD">2. getter and setter</h3>

```python
class myClass:
    def __init__(self):
        self.name = ""
        self.set_name()
        self.get_name()

    def set_name(self): # setter method
        name = input("what is your name :")
        self.name = name

    def get_name(self): # getter method
        print(f"Hello, {self.name}")

class1 = myClass()
```

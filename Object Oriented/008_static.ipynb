<h2 style="color:#C83F49"> Static Method </h2>

```python
# static method is a method that belongs to a class rather than an instance of the class

class Adder:
    def __init__(self,n):
        self.num = n

    def add_to(self,x):
        self.num += x

    # does not need self because it does not need class instance to work
    # can be used with or without instance
    # used so it can be shipped with a class
    @staticmethod
    def add(a,b):
        return a+b

num1 = Adder(10)
print(num1.num)
num1.add_to(5)
print(num1.num)

print(num1.add(4,4))    # using instance naming
print(Adder.add(5,5))   # using Class naming
```

<h2 style="color:#C83F49"> Inheritance </h2>

```python
# it allows a new class subclass to inherit properties and behaviors from an existing class or superclass
```

```python
# Parent Class
class Employee:
    def __init__(self):
        a = input("Enter your name : ")
        b = input("What is your role : ")
        c = input("What is your Employee ID : ")
        self._name = a
        self._role = b
        self._ID = c

    @property
    def employee(self):
        return f"your name is {self._name} and you are a {self._role} having ID {self._ID}"

# Child Class
class Language(Employee):
    def sel_lan(self):
        d = input("What language do you use : ")
        self._language = d

Aaditya = Employee()
print(Aaditya.employee)

# Child Class can call Parent Class methods but not vice versa 
# Parent Class is automatically invoked
Aryan = Language()
print(Aryan.employee)
Aryan.sel_lan()
```

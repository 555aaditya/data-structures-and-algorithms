<h1 style="color:#fc0000"> 118. Pascal Triangle</h1>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n&sup2)</h3>

```python
def generator(n: int) -> list[list[int]]:
    res = [[1]]
    for i in range(n-1):
        temp = [0] + res[-1] + [0]
        row = []
        for j in range(len(res[-1])+1):
            row.append(temp[j]+temp[j+1])
        res.append(row)
    return res

if __name__ == '__main__':
    numRows = 6
    print(generator(numRows))
```

<h1 style="color:#fc0000"> 229. Majority Element II</h1>

<h3 style="color:#cccccc"> Brute Force</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def majority(a: list[int]) -> list[int]:
    n = len(a)//3
    res = set()
    for num1 in a:
        count = 0
        for num2 in a:
            if num1 == num2:
                count += 1
        if count > n:
            res.add(num1)
    return res

if __name__ == '__main__':
    a = [1,1,1,2,2,3,3,3]
    print(majority(a))
```

<h3 style="color:#cccccc"> HashMap</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def majority(arr: list[int]) -> list[int]:
    n = len(arr)//3
    hashMap = {}
    res = []
    for num in arr:
        hashMap[num] = hashMap.get(num,0) + 1
        if hashMap[num] > n:
            res.append(num)
    return res

if __name__ == '__main__':
    a = [1,1,1,2,2,3,3,3]
    print(majority(a))
```

<h3 style="color:#cccccc"> Moore Algorithm</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def majority(arr: list[int]) -> list[int]:
    count1, count2, candidate1, candidate2 = 0, 0, None, None
    res = []
    n = len(arr)//3

    for num in arr:
        if num == candidate1:
            count1 += 1
        elif num == candidate2:
            count2 += 1
        elif count1 == 0:
            candidate1, count1 = num, 1
        elif count2 == 0:
            candidate2, count2 = num, 1
        else:
            count1 -= 1
            count2 -= 1

    for num in [candidate1, candidate2]:
        if arr.count(num) > n:
            res.append(num)

    return res

if __name__ == '__main__':
    a = [1,1,1,2,2,3,3,3]
    print(majority(a))
```

<h3 style="color:#cccccc"> Hash Map</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n+k)</h3>

```python
def threeSum(arr: list[int]) -> list[list[int]]:
    n = len(arr)
    res = set()

    for i in range(n):
        hashMap = {}
        for j in range(i+1,n):
            target = -(arr[i]+arr[j])
            if target in hashMap:
                triplet = tuple(sorted([arr[i],arr[j],target]))
                res.add(triplet)
            hashMap[arr[j]] = 1

    return list(res)

if __name__ == '__main__':
    a = [-1,0,1,2,-1,-4]
    print(threeSum(a))
```

<h3 style="color:#cccccc"> Sorted Approach</h3>

<h3 style="color:#cccccc"> Time Complexity - O(nlogn)+O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(k)</h3>

```python
def threeSum(nums: list[int]) -> list[list[int]]:
    n = len(nums)
    res = []
    nums.sort()

    for i in range(n):
        if i == 0 or nums[i] != nums[i-1]:
            j = i+1
            k = n-1
            while k>j:
                sum = nums[j] + nums[k]
                if sum == -nums[i]:
                    res.append([nums[i], nums[j], nums[k]])
                    j += 1
                    while(j<k and nums[j] == nums[j-1]):
                        j+= 1
                elif sum > -nums[i]:
                    k -= 1
                elif sum < -nums[i]:
                    j += 1
        
    return res

if __name__ == '__main__':
    a = [-1,0,1,2,-1,-4]
    print(threeSum(a))
```

<h3 style="color:#cccccc"> Hash Map</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup3)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def fourSum(arr: list[int], target: int) -> list[list[int]]:
    n = len(arr)
    res = set()

    for i in range(n):
        for j in range(i+1,n):
            hashMap = {}
            for k in range(j+1,n):
                ans = target - (arr[i]+arr[j]+arr[k])
                if ans in hashMap:
                    temp = tuple(sorted((arr[i],arr[j],arr[k],ans)))
                    res.add(temp)
                hashMap[arr[k]] = 1
                    
    return [list(t) for t in res]

if __name__ == '__main__':
    a = [1,0,-1,0,-2,2]
    t = 0
    print(fourSum(a,t))
```

<h3 style="color:#cccccc"> Two Pointer</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup3)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def fourSum(arr: list[int], target: int) -> list[list[int]]:
    n = len(arr)
    res = []
    arr.sort()

    for i in range(n):
        
        if i>0 and arr[i] == arr[i-1]:
            continue
            
        for j in range(i+1,n):
            
            if j>(i+1) and arr[j] == arr[j-1]:
                continue

            left = j+1
            right = n-1

            while left < right:
                sum = arr[i]+arr[j]+arr[left]+arr[right]

                if sum == target:
                    res.append([arr[i],arr[j],arr[left],arr[right]])
                    left+=1
                    right-=1
                    while left<right and arr[left] == arr[left-1]:
                        left+=1
                    while left<right and arr[right] == arr[right+1]:
                        right-=1

                elif sum < target:
                    left+=1

                else:
                    right-=1

    return res

if __name__ == '__main__':
    a = [1,0,-1,0,-2,2]
    t = 0
    print(fourSum(a,t))
```

<h1 style="color:#fc0000"> 594. Longest Harmonious Subsequence</h1>

<h3 style="color:#cccccc"> Brute Force</h3>

```python
def findLHS(nums: list[int]) -> int:
        max_len = 0
        for i in range(len(nums)):
            ans = []
            has_pair = False
            for num in nums:
                if num == nums[i]:
                    ans.append(num)
                if num == nums[i]+1:
                    ans.append(num)
                    has_pair = True
            if has_pair:
                max_len = max(max_len, len(ans))
        return max_len 

if __name__ == '__main__':
    a = [1,3,2,2,5,2,3,7]
    print(findLHS(a))
```

<h3 style="color:#cccccc"> Hash Map</h3>

```python
def findLHS(nums: list[int]) -> int:
    hashMap = {}
    for num in nums:
        hashMap[num] = hashMap.get(num, 0) + 1

    max_len = 0
    for i in nums:
        if i+1 in nums:
            max_len = max(max_len, hashMap[i]+hashMap[i+1])

    return max_len

if __name__ == '__main__':
    a = [1,3,2,2,5,2,3,7]
    print(findLHS(a))
```

<h3 style="color:#cccccc"> Sorted Approach</h3>

```python
def findLHS(nums: list[int]) -> int:
    nums.sort()
    start = 0
    max_len = 0

    for end in range(len(nums)):
        while nums[end] - nums[start] > 1:
            start += 1
        if nums[end] - nums[start] == 1:
            max_len = max(max_len, end-start+1)
    return max_len

if __name__ == '__main__':
    a = [1,3,2,2,5,2,3,7]
    print(findLHS(a))
```

<h1 style="color:#fc0000"> 3304. Find the K-th Character in String Game I</h1>

```python
def kthCharacter(k: int) -> str:
    word = "a"

    while len(word) <= k:
        temp = word
        
        for char in word:
            if char == 'z':
                temp += 'a'
            else:
                temp += chr(ord(char) + 1)
            
        word = temp      

    return word[k - 1]

if __name__ == '__main__':
    k = 4
    print(kthCharacter(k))
```

<h1 style="color:#fc0000"> 1394. Find Lucky Integer in an Array</h1>

```python
def findLucky(arr: list[int]) -> int:
    freq = {}
    _max = -1

    for num in arr:
        freq[num] = freq.get(num,0) + 1

    for num in freq:
        if freq[num] == num:
            _max = max(num,_max)

    return _max

if __name__ == '__main__':
    arr = [1,2,2,3,3,3]
    print(findLucky(arr))
```

```python
def findLucky(arr: list[int]) -> int:
    arr.sort(reverse=True)
    _max = -1
    n = len(arr)
    i = 0

    while i < n:
        current = arr[i]
        count = 0
        while i < n and arr[i] == current:
            count += 1
            i += 1
        if count == current:
            _max = max(_max, current)
        
    return _max

if __name__ == '__main__':
    arr = [1,2,2,3,3,3]
    print(findLucky(arr))
```

<h1 style="color:#fc0000"> 12. Longest subarray with sum K</h1>

<h3 style="color:#cccccc"> Brute Force Approach - Create all possible subarrays and check sum</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def longestSubarray(arr: list[int], k: int) -> int:
    maximum = 0

    for i in range(len(arr)):
        sum_ = 0
        for j in range(i,len(arr)):
            sum_ += arr[j]
            if sum_ == k:
                maximum = max(maximum, j-i+1)
    return maximum

if __name__ == '__main__':
    arr = [1, 2, 3, 1, 1, 1, 1, 2, 3]
    k = 6
    print(longestSubarray(arr, k))
```

<h3 style="color:#cccccc"> Prefix Sum Approach </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def longestSubarray(arr: list[int], k: int) -> int:
    prefix = 0
    maximum = 0
    hash_set = {}

    for i in range(len(arr)):
        prefix += arr[i]

        if prefix == k:
            maximum = i + 1

        if (prefix - k) in hash_set:
            maximum = max(maximum, i - hash_set[prefix-k])

        if prefix not in hash_set:
            hash_set[prefix] = i

    print(hash_set)
    return maximum
    
if __name__ == '__main__':
    arr = [1, 2, 3, 1, 1, 1, 1, 2, 3]
    k = 6
    print(longestSubarray(arr, k))
```

<h1 style="color:#fc0000"> 12. Count of subarray whose sum is K</h1>

<h3 style="color:#cccccc"> Prefix Sum Approach </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def countSubarray(arr: list[int], k: int) -> int:
    prefix = 0
    count = 0
    hash_set = {0:1}

    for i in range(len(arr)):
        prefix += arr[i]

        if (prefix-k) in hash_set:
            count += hash_set[prefix-k]

        hash_set[prefix] = hash_set.get(prefix,0) + 1

    return count

if __name__ == '__main__':
    arr = [1, 2, 3, 1, 1, 1, 1, 2, 3]
    k = 6
    print(countSubarray(arr, k))
```

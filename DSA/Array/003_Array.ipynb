<h1 style="color:#fc0000"> 12. Longest subarray with sum K</h1>

<h3 style="color:#cccccc"> Brute Force Approach - Create all possible subarrays and check sum</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def longestSubarray(arr: list[int], k: int) -> int:
    maximum = 0

    for i in range(len(arr)):
        sum_ = 0
        for j in range(i,len(arr)):
            sum_ += arr[j]
            if sum_ == k:
                maximum = max(maximum, j-i+1)
    return maximum

if __name__ == '__main__':
    arr = [1, 2, 3, 1, 1, 1, 1, 2, 3]
    k = 6
    print(longestSubarray(arr, k))
```

<h3 style="color:#cccccc"> Prefix Sum Approach </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def longestSubarray(arr: list[int], k: int) -> int:
    prefix = 0
    maximum = 0
    hash_set = {}

    for i in range(len(arr)):
        prefix += arr[i]

        if prefix == k:
            maximum = i + 1

        if (prefix - k) in hash_set:
            maximum = max(maximum, i - hash_set[prefix-k])

        if prefix not in hash_set:
            hash_set[prefix] = i

    print(hash_set)
    return maximum
    
if __name__ == '__main__':
    arr = [1, 2, 3, 1, 1, 1, 1, 2, 3]
    k = 6
    print(longestSubarray(arr, k))
```

<h1 style="color:#fc0000"> 13. Count of subarray whose sum is K</h1>

<h3 style="color:#cccccc"> Prefix Sum Approach </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def countSubarray(arr: list[int], k: int) -> int:
    prefix = 0
    count = 0
    hash_set = {0:1}

    for i in range(len(arr)):
        prefix += arr[i]

        if (prefix-k) in hash_set:
            count += hash_set[prefix-k]

        hash_set[prefix] = hash_set.get(prefix,0) + 1

    return count

if __name__ == '__main__':
    arr = [1, 2, 3, 1, 1, 1, 1, 2, 3]
    k = 6
    print(countSubarray(arr, k))
```

<h1 style="color:#fc0000"> 14. Two Sum</h1>

<h3 style="color:#cccccc"> Brute Force Approach </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def twoSum(arr: list[int], k: int) -> list[int]:
    for i in range(len(arr)):
        for j in range(i+1,len(arr)):
            if (i == j):
                continue
            if (arr[i]+arr[j] == k):
                return [i,j]

if __name__ == '__main__':
    arr = [2,7,11,15]
    k = 18
    twoSum = twoSum(arr,k)
    print(twoSum)
```

<h3 style="color:#cccccc"> Hash Set </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def twoSum(arr: list[int], k: int) -> list[int]:
    hash_set = {}
    for i in range(len(arr)):
        if (k-arr[i]) in hash_set:
            return [hash_set[k-arr[i]],i]

        if arr[i] not in hash_set:
            hash_set[arr[i]] = i

if __name__ == '__main__':
    arr = [2,7,11,15]
    k = 18
    twoSum = twoSum(arr,k)
    print(twoSum)
```

<h3 style="color:#cccccc"> 2 Pointer</h3>

<h3 style="color:#cccccc"> Time Complexity - O(nlogn)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def twoSum(arr: list[int], k: int) -> list[int]:
    arr.sort()
    left, right = 0, (len(arr)-1)

    while left < right:
        sum_ = (arr[left] + arr[right])
        if (sum_ == k):
            return [arr[left], arr[right]]
        elif sum_ < k:
            left+=1
        elif sum_ > k:
            right-=1
    return []

if __name__ == '__main__':
    arr = [2,7,11,15]
    k = 18
    twoSum = twoSum(arr,k)
    print(twoSum)
```

<h1 style="color:#fc0000"> 15. Sort an array of 0's, 1's and 2's</h1>

<h3 style="color:#cccccc"> Brute Force Approach - Sorting Algorithm </h3>

```python
def quickSort(arr: list[int]) -> list[int]:
    
    if len(arr) <= 1:
        return arr  
        
    pivot = arr[len(arr) // 2] 
    left = [x for x in arr if x < pivot]  
    middle = [x for x in arr if x == pivot] 
    right = [x for x in arr if x > pivot]  
    
    return quickSort(left) + middle + quickSort(right)

arr = [2, 0, 2, 1, 1, 0]
print(quickSort(arr))
```

<h3 style="color:#cccccc"> Frequency Count Approach</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def sortArray(arr: list[int]) -> list[int]:
    red, blue, white = 0, 0, 0

    for j in arr:
        if j == 0:
            red += 1
            
        elif j == 1:
            blue += 1

        elif j == 2:
            white += 1

    for i in range(len(arr)):
        if red > 0:
            arr[i] = 0
            red -= 1

        elif blue > 0:
            arr[i] = 1
            blue -= 1

        elif white > 0:
            arr[i] = 2
            white -= 1

    return arr

arr = [2, 0, 2, 1, 1, 0]
print(sortArray(arr))
```

<h3 style="color:#cccccc"> Dutch National Flag Algorithm</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def dutchNational(arr: list[int]) -> list[int]:
    low, mid, high = 0, 0, len(arr)-1

    while mid <= high:
        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low+=1
            mid+=1

        elif arr[mid] == 1:
            mid+=1

        elif arr[mid] == 2:
            arr[mid], arr[high] = arr[high], arr[mid]
            high-=1

    return arr


arr = [2, 0, 2, 1, 1, 0]
print(dutchNational(arr))
```

<h1 style="color:#fc0000"> 16. Majority element in an array</h1>

<h3 style="color:#cccccc"> Brute Force Approach  </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def majorityElement(arr: list[int]) -> int:
    for i in range(len(arr)): 
        count = 0
        for j in range(len(arr)):
            if (arr[j] == arr[i]):
                count += 1

        if (count) > (len(arr)//2):
            return arr[i]
    return -1

arr = [5,2,3,5,3,5,5,5,4,5,2]
print(majorityElement(arr))
```

<h3 style="color:#cccccc"> Hash Map - Frequency Count </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def majorityElement(arr: list[int]) -> int:
    freq = {}

    for num in arr:
        freq[num] = freq.get(num,0) + 1
        if freq[num] > (len(arr)//2):
            return num
    return -1

arr = [5,2,3,5,3,5,5,5,4,5,2]
print(majorityElement(arr))
```

<h3 style="color:#cccccc"> Moore Voting Algorithm </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>
<h3 style="color:#cccccc"> Final candidate is the majority element because majority element always survives cancellation </h3>

```python
def majorityElement(arr: list[int]) -> int:
    candidate, count = None, 0

    for num in arr:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)
    return candidate
    
arr = [5,2,3,5,3,5,5,5,4,5,2]
print(majorityElement(arr))
```

<h1 style="color:#fc0000"> 17. Kadane Algorithm</h1>

<h3 style="color:#cccccc"> Brute Force Approach  </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup3)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def kadane(arr: list[int]) -> int:
    maximum = float('-inf')

    for i in range(len(arr)):
        for j in range(i,len(arr)):
            sum = 0
            for k in range(i,j+1):
                sum += arr[k]
            maximum = max(sum, maximum)
    return maximum

arr = [1,2,-1,-3,5,4,-6,2,5,9]
print(kadane(arr))
```

<h3 style="color:#cccccc"> Kadane Algorithm  </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

<h3 style="color:#cccccc"> negative sum won't contribute to max  </h3>

```python
def kadane(arr: list[int]) -> int:
    maximum = float('-inf')
    currentSum = 0

    for num in arr:
        currentSum += num
        maximum = max(currentSum, maximum)

        if currentSum < 0:
            currentSum = 0
    return maximum

arr = [1,2,-1,-3,5,4,-6,2,5,9]
print(kadane(arr))
```

<h1 style="color:#fc0000"> 18. Rearrange the array in alternating positive and negative items</h1>

<h3 style="color:#cccccc"> Brute Force Approach  </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def rearrange(arr: list[int]) -> list[int]:
    n = len(arr)
    result = []
    pos = []
    neg = []

    for num in arr:
        if num < 0:
            neg.append(num)

    for num in arr:
        if num >= 0:
            pos.append(num)

    i = j = 0
    while i < len(pos) and j < len(neg):
        result.append(pos[i])
        result.append(neg[j])
        i+=1
        j+=1
        
    return result
 
arr = [1,2,-1,-3,5,4,-6,-2,-5,9]
print(rearrange(arr))
```

<h3 style="color:#cccccc"> Two Pointer Approach  </h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def rearrange(arr: list[int]) -> list[int]:
    n = len(arr)
    i, j = 0, 1

    while i < n and j < n:
        while i < n and arr[i] >= 0:  
            i += 2
        while j < n and arr[j] < 0:  
            j += 2
        
        if i < n and j < n:  
            arr[i], arr[j] = arr[j], arr[i]
        
    return arr
    
 
arr = [1,2,-1,-3,5,4,-6,-2,-5,9]
print(rearrange(arr))
```

<h1 style="color:#fc0000"> 19. Stock buy and sell</h1>

```python
def maxProfit(arr: list[int]) -> int:
    profit_max = 0
    min_ = float('inf')

    for num in arr:
        min_ = min(num,min_)
        profit_max = max(num-min_,profit_max)
        
    return profit_max

maxProfit([7,1,5,3,6,4])
```

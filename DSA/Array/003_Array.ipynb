<h1 style="color:#fc0000"> 12. Longest subarray with sum K</h1>

<h3 style="color:#cccccc"> Brute Force Approach - Create all possible subarrays and check sum</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def longestSubarray(arr: list[int], k: int) -> int:
    maximum = 0

    for i in range(len(arr)):
        sum_ = 0
        for j in range(i,len(arr)):
            sum_ += arr[j]
            if sum_ == k:
                maximum = max(maximum, j-i+1)
    return maximum

if __name__ == '__main__':
    arr = [1, 2, 3, 1, 1, 1, 1, 2, 3]
    k = 6
    print(longestSubarray(arr, k))
```

<h1 style="color:#d9d200"> Binary Search </h1>

<h1 style="color:#fc0000"> 704. Binary Search</h1>

<h3 style="color:#cccccc"> Iterative</h3>
<h3 style="color:#cccccc"> Time Complexity - O(logn)</h3>

```python
def binarySearch(arr: list[int], target: int) -> int:
    low, high = 0, len(arr)-1
    while low <= high:
        mid = (low+high)//2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            low = mid+1

        else:
            high = mid-1

if __name__ == '__main__':
    arr = [-1,0,3,5,9,12]
    target = 9
    print(binarySearch(arr, target))
```

<h3 style="color:#cccccc"> Recursive</h3>
<h3 style="color:#cccccc"> Time Complexity - O(logn)</h3>

```python
def binarySearch(arr: list[int], target: int, low: int, high: int) -> int:
    if low > high:
        return -1

    mid = (low+high)//2

    if (arr[mid] == target):
        return mid

    elif arr[mid] < target:
        return binarySearch(arr, target, mid+1, high)

    else:
        return binarySearch(arr, target, low, mid-1)

if __name__ == '__main__':
    arr = [-1,0,3,5,9,12]
    target = 9
    low = 0
    high = len(arr)-1
    print(binarySearch(arr, target, low, high))
```

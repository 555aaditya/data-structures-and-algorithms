<h2 style="color:#fc0000"> 1. Printing an input N times </h2>

```python
i = 0
def printName(name:str,n:int) -> str:
    global i
    if (i == n):
        return
    else:
        print(i+1,name)
        i+=1
        printName(name,n)

if __name__ == "__main__":
    x = input("Enter your name : ")
    n = int(input("How many times : "))
    printName(x,n)
```

```python
def printName(name:str,n:int,i:int) -> str:
    if i == n:
        return
    print(i+1,name)
    printName(name,n,i+1)

if __name__ == "__main__":
    x = input("Enter your name : ")
    n = int(input("How many times : "))
    printName(x,n,0)
```

<h2 style="color:#fc0000"> 2. Sum of first N numbers </h2>

```python
i = 0
total = 0

def N_Sum(n: int) -> int:
    global i
    global total
    if i > n:
        return
    total += i
    i += 1
    N_Sum(n)

N_Sum(11)
print(total)
```

```python
def N_Sum(n:int) -> int:
    if (n==0):
        return 0
    return n + N_Sum(n-1)

print(N_Sum(11))
```

<h2 style="color:#fc0000"> 3. Reversing an Array </h2>

```python
def swap(a:list,i:int,j:int):
    a[i],a[j] = a[j],a[i]

def reverse(a,i,n):
    if (i>=n/2):
        return 
    swap(a,i,n-i-1)
    reverse(a,i+1,n)

if __name__ == "__main__":
    a = [1,2,3,4,5]
    n = len(a)
    reverse(a,0,n)
    print(a)
```

<h2 style="color:#fc0000"> 4. Checking for Palindrome </h2>

```python
def palindrome(a:str,i:int,n:int) -> bool:
    if (i>=n/2):
        return True
    if (a[i]!=a[n-i-1]):
        return False
    return palindrome(a,i+1,n)

if __name__ == "__main__":
    a = "MADAM"
    n = len(a)
    print(palindrome(a,0,n))
```

<h2 style="color:#fc0000"> 5. Fibonacci using multiple recursion calls </h2>

```python
def fibonacci(n:int) ->  int:
    if (n<=1):
        return n
    last = fibonacci(n-1)
    slast = fibonacci(n-2)
    return last + slast 

fibonacci(5)
```

<h2 style="color:#fc0000">1. Subsequences using for loop </h2>

```python
# using for loop the program is more complex

def powerSet(s):
    sets = [[]] 
    
    for i in s:
        new_sets = [] 
        for subset in sets:
            new_set = subset + [i]  
            new_sets.append(new_set)  
        sets.extend(new_sets)
    return sets

input_set = ['a', 'b', 'c']
all_subsets = powerSet(input_set)
for subset in all_subsets:
    print(subset)
```

```python
# for all the indexes we can either take or not take an element

def powerSet(s):
    sets = [[]]
    
    for i in s:
        newsets = [k + [i] for k in sets]
        sets.extend(newsets)
    return sets

input_set = ['a', 'b', 'c']
all_subsets = powerSet(input_set)
for subset in all_subsets:
    print(subset)
```

<h2 style="color:#fc0000">2. Subsequences using Recursion </h2>

<h3 style="color:#cccccc">The code essentially implements a recursive backtracking algorithm to generate all possible subsets of a given array arr. It explores two possibilities at each step: including the current element in the subset or excluding it, leading to an exponential time complexity proportional to 2^N, where N is the length of the input array arr</h3>

```python
# for arrays

power_set = []
def powerset(i,temp_arr,n,arr):
    global power_set
    if (i>=n):
        power_set.append(temp_arr[:])
        return
    temp_arr.append(arr[i])
    powerset(i+1,temp_arr,n,arr)
    temp_arr.pop()
    powerset(i+1,temp_arr,n,arr)

arr = [1,2,3,4]
n = len(arr)
powerset(0,[],n,arr)

for i in power_set:
    print(i)
```

```python
# shorter method for arrays

def print_subsequences_recursive(arr, index, subsequence):
    if index == len(arr):
        print(subsequence)
        return
    # Include current element
    print_subsequences_recursive(arr, index + 1, subsequence + [arr[index]])
    # Exclude current element
    print_subsequences_recursive(arr, index + 1, subsequence)

arr = [1, 2, 3]
print_subsequences_recursive(arr, 0, [])
```

```python
# using backtracking 

def print_subsequences_backtracking(arr, index, subsequence):
    print(subsequence)
    for i in range(index, len(arr)):
        subsequence.append(arr[i])
        print_subsequences_backtracking(arr, i + 1, subsequence)
        subsequence.pop()

arr = [1, 2, 3]
print_subsequences_backtracking(arr, 0, [])
```

```python
# for Strings

power_set = []
def powerSet(i,n,temp_str,string):
    global power_set
    if (i>=n):
        power_set.append(temp_str)
        return
    temp_str += string[i]
    powerSet(i+1,n,temp_str,string)
    temp_str = temp_str[:-1]
    powerSet(i+1,n,temp_str,string)
    
powerSet(0,len("abcd"),"","abcd")    
for i in power_set:
    print(f"\"{i}\"")
```

<h2 style="color:#fc0000">3. Subsequences following a condition </h2>

```python
# which has a particular sum

power_set = []
def powerSet(i,n,temp_arr,arr,sum,s):
    global power_set
    if (i>=n):
        if (s==sum):
            power_set.append(temp_arr[:])
        return
    temp_arr.append(arr[i])
    s+=arr[i]
    powerSet(i+1,n,temp_arr,arr,sum,s)
    temp_arr.pop()
    s-=arr[i]
    powerSet(i+1,n,temp_arr,arr,sum,s)

powerSet(0,len([1,2,3,4,5,6]),[],[1,2,3,4,5,6],6,0)

for _ in power_set:
    print(_)
```

```python
# which has vowel in start and consonant at the end 

power_set = []

def checkCondition(S:str) -> bool:
    if not S:
        return False
    if (S[0] in "aeiou") and (S[-1] not in "aeiou"):
        return True
    return False

def powerSet(i,n,string,temp_str):
    global power_set
    if (i>=n):
        if checkCondition(temp_str):
            power_set.append(temp_str[:])
        return
    temp_str += string[i]
    powerSet(i+1,n,string,temp_str)
    temp_str = temp_str[:-1]
    powerSet(i+1,n,string,temp_str)

powerSet(0,len("abed"),"abed","")

for i in power_set:
    print(i)
```

```python
# number of subsequences which meet the conditiion

def powerSet(i,n,s,sum,arr):
    if (i>=n):
        if (s==sum): return 1
        else: return 0

    s+=arr[i]
    l = powerSet(i+1,n,s,sum,arr)
    s-=arr[i]
    r = powerSet(i+1,n,s,sum,arr)
    return l+r

powerSet(0,len([1,2,3,4]),0,2,[1,2,3,4])
```

<h2 style="color:#fc0000">4. Subsequence Game</h2>

<a href="https://www.geeksforgeeks.org/problems/string-subsequence-game5515/1" style="color:#cccccc; font-size: 20px;">https://www.geeksforgeeks.org/problems/string-subsequence-game5515/1</a>


```python
class Solution:
    def checkCondition(self,s:str) -> bool:
        if s[0] in 'aeiou' and s[-1] not in 'aeiou':
            return True
        return False
        
    def powerSet(self,i:int,temp:str,s:str,result:set) -> set:
        if (i==len(s)):
            if temp and self.checkCondition(temp):
                result.add(temp)
            return
        
        temp += s[i]
        self.powerSet(i+1,temp,s,result)
        temp = temp[:-1]
        self.powerSet(i+1,temp,s,result)
        
    def allPossibleSubsequences (self, S:str) -> set:
        result = set()
        self.powerSet(0,"",S,result)
        return sorted(result)




if __name__ == '__main__':
    S = "abed"
    ans=set()
    ob = Solution()
    ans = ob.allPossibleSubsequences(S)
    for i in ans:
        print(i,end=" ")
        print()
```

<h2 style="color:#fc0000">5. When the element can be repeated in a subsequence to meet a condition</h2>

<a href="https://leetcode.com/problems/combination-sum/" style="color:#cccccc; font-size: 20px;">https://leetcode.com/problems/combination-sum/</a>

```python
def powerSet(i,arr,target,temp,result):
    if (i==len(arr)):
        if(target == 0):
            result.append(temp[:])
        return
        
    temp.append(arr[i])
    if(target>=0):
        powerSet(i,arr,target-arr[i],temp,result)
    temp.pop()
    powerSet(i+1,arr,target,temp,result)


solution = []
powerSet(0,[1,2,3],4,[],solution)
print(solution)
```

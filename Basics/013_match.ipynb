<h2 style="color:#C83F49"> Match Case </h2>

```python
def match_example(value):
    match value:
        case 1:
            print("The value is 1")
        case 2:
            print("The value is 2")

        # default case
        case _:
            print("The value is something else")

match_example(1)  
match_example(2)  
match_example(3)  
```

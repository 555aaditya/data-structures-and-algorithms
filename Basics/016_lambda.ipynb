<h2 style="color:#C83F49"> Lambda Functions </h2>

```python
# Lambda functions are small, anonymous functions that can have any number of arguments but only one expression.
# x = lambda arguments: expression

add = lambda x, y: x + y
print(add(3, 4)) 
```

```python
# to return a function as an arguement we can use lambda so we don't have to create a function

def operate(func, x, y):
    return func(x, y)

result = operate((lambda x, y: x * y), 3, 4)
print(result)  
```

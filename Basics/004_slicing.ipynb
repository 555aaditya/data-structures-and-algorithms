<h2 style="color:#C83F49"> String Slicing </h2>

<h3 style="color:#C1A1BD">1. One Arguement</h3>

```python
# will return the single character at that index

x = "Aaditya Sinha"

single_character = x[6]
print(single_character)
```

<h3 style="color:#C1A1BD">2. Two Arguements</h3>

```python
# will return the string from the start to the end index

x = "Aaditya Sinha"

first_name = x[0:7]
last_name = x[8:]
print(first_name)
print(last_name)
```

<h3 style="color:#C1A1BD">3. Three Arguements</h3>

```python
# used to add step size to string slicing

x = "AxaxdxixtxyxaxSxixnxhxax"
y = "Aaditya"

name = x[::2]
reverse = y[::-1]
print(name)
print(reverse)
```

<h3 style="color:#C1A1BD">4. Repeating Slicing</h3>

```python
# used like a function

website1 = "https://www.google.com"
website2 = "https://www.github.com"
website3 = "https://www.linkedin.com"

slice1 = slice(12,-4)

print(website1[slice1])
print(website2[slice1])
print(website3[slice1])
```

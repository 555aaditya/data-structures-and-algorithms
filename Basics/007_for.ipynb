<h2 style="color:#C83F49"> For Loops </h2>

```python
# will run the code block a limited amount of times

for i in range(1,11):
    print(i)
    
print("\n")

for i in "Aaditya":
    print(i,end="")

print("\n")

for i in range(1,6):
    for j in range(i):
        print("*",end="")
    print("")
```

<h2 style="color:#C83F49"> Tuples </h2>

```python
# tuple is an ordered, immutable collection of elements
# if the elements themselves are mutable they can be modified
```

<h3 style="color:#C1A1BD">1. Initialization</h3>

```python
# Empty tuple
empty_tuple = ()
print(empty_tuple)

# Tuple with elements
my_tuple = (1, 2, 3, 4, 5)
print(my_tuple)

# Tuple with mixed data types
mixed_tuple = ('apple', 3.14, True, [1, 2, 3])
print(mixed_tuple)

# Nested tuple
nested_tuple = ('hello', (1, 2, 3), ['a', 'b', 'c'])
print(nested_tuple)

# Single-element tuple (note the comma)
single_tuple = (42,) 
print(single_tuple)
```

<h3 style="color:#C1A1BD">2. Accessing</h3>

```python
my_tuple = (1, 2, 3, 4, 5)
print(my_tuple[0])   
print(my_tuple[-1],'\n')


my_tuple1 = (1, 2, 3)
for item in my_tuple1:
    print(item)
```

<h3 style="color:#C1A1BD">3. Slicing</h3>

```python
my_tuple = (1, 2, 3, 4, 5)
print(my_tuple[1:3])  
```

<h3 style="color:#C1A1BD">4. Concatenation</h3>

```python
tuple1 = (1, 2, 3)
tuple2 = ('a', 'b', 'c')
concatenated_tuple = tuple1 + tuple2
print(concatenated_tuple)
```

<h3 style="color:#C1A1BD">5. Membership</h3>

```python
my_tuple = (1, 2, 3)
print(1 in my_tuple)
print('a' not in my_tuple)  
```

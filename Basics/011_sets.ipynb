<h2 style="color:#C83F49"> Sets </h2>

```python
# sets are mutable objects
# after a set is created, you can modify it by adding or removing elements 
# the contents of a set cannot be modified through item assignment, as sets do not support indexing
```

<h3 style="color:#C1A1BD">1. Initialization</h3>

```python
# sets are similar to the sets in maths
# set is an unordered collection of unique elements

# using curly braces
set1 = {1,2,3,4,5}
print(set1,'\n')

# using set constructor
set2 = set([1,2,3,2,1])
print(set2,'\ndoes not repeat elements\n')

# empty set
set3 = set()
print(set3,'\n')
```

<h3 style="color:#C1A1BD">2. Add Element</h3>

```python
# adding a single element

set1 = {1,2,3}
print(set1,"\nbefore adding an element\n")
set1.add(4)
print(set1,"\nafter adding an element")

# adding multiple elements usng iterable

set2 = {1,2}
print(set2,"\nbefore adding elements\n")
set2.update([3,4,5])
print(set2,"\nAfter adding elements\n")
set2.update({6,7,8})
print(set2,"\nAfter adding elements\n")
```

<h3 style="color:#C1A1BD">3. Removing Elements</h3>

```python
# remove()
# if the element is not present, it raises a KeyError

set1 = {1,2,3,4}
print(set1,"\nbefore removing an element\n")
set1.remove(3)
print(set1,"\nafter removing an element\n")

# discard()
# If the element is not present, it does nothing.

set2 = {1,2,3,4}
print(set2,"\nbefore removing an element\n")
set2.discard(3)
print(set2,"\nafter removing an element\n")

# pop()
# removes and returns an arbitrary element from the set
# if the set is empty, it raises a KeyError

set3 = {1,2,3,4}
print(set3,"\nbefore removing an element\n")
x = set3.pop()
print(set3,"\nafter removing an element\n")
print(f"element popped : {x}\n")

# clear()
# removes all elements from the set, making it empty.

set4 = {1,2,3,4}
print(set4,"\nbefore clearing set\n")
set4.clear()
print(set4,"\nafter clearing\n")
```

<h3 style="color:#C1A1BD">4. Set Operations</h3>

```python
# union 

set1 = {1,2,3}
set2 = {3,4,5}
print(f"set1 is {set1} and set2 is {set2}\n")

set3 = set1 | set2
print(f"union of set1 and set2 is {set3}\n")

# intersection

set4 = set1 & set2
print(f"intersection of set1 and set2 is {set4}\n")

# difference
# Returns a new set containing elements present in the first set but not in the second set

set5 = {1,2,3,4}
set6 = {3,4,5,6}

diff_set = set5 - set6
print(f"difference of {set5} and {set6} is {diff_set}\n")
print(f"difference of {set6} and {set5} is {set6 - set5}\n")

# symmetric difference
# Returns a new set containing elements present in either set, but not in both sets

set7 = {1,2,3,4}
set8 = {3,4,5,6}

print(f"symmetric difference of {set7} and {set8} is {set7 ^ set8}\n")

# subset <=
# Returns True if all elements of the first set are present in the second set

set9 = {1,2,3}
set10 = {1,2,3,4,5,6}

print(f"is {set9} a subset of {set10} : {set9 <= set10}\n")

# proper subset <
# if all elements of the first set are present in the second set, and the sets are not equal

set9 = {1,2,3}
set10 = {1,2,3,4,5,6}

print(f"is {set9} a proper subset of {set9} : {set9 < set9}\n")
print(f"is {set9} a proper subset of {set10} : {set9 < set10}\n")

# superset >=
# if all elements of the second set are present in the first set.

print(f"is {set10} a superset of {set9} : {set10 >= set9}\n")

# proper superset >
# if all elements of the second set are present in the first set, and the sets are not equal

print(f"is {set10} a proper superset of {set9} : {set10 > set9}\n")
print(f"is {set9} a proper superset of {set9} : {set9 > set9}\n")

# disjoint
# if the two sets have no common elements

set11 = {1,2,3}
set12 = {4,5}

print(f"is {set9} and {set10} disjoint : {set9.isdisjoint(set10)}\n")
print(f"is {set11} and {set12} disjoint : {set11.isdisjoint(set12)}\n")
```

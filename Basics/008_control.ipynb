<h2 style="color:#C83F49"> Loop Control Statements </h2>

<h3 style="color:#C1A1BD">1. Break</h3>

```python
# it terminates the remaining loop entirely when it's condition is met

for i in "123456789":
    print(i)
    if i == "4":
        break
```

<h3 style="color:#C1A1BD">2. Continue</h3>

```python
# it skips the current iteration and continues the rest of the interations

for i in "123456789":
    if i == "4" or i == "6":
        continue
    print(i)
```

<h3 style="color:#C1A1BD">3. Pass</h3>

```python
# In Python, the pass statement is a null operation. It doesn't do anything when executed

for i in range(21):
    if i == 13:
        pass
    else:
        print(i,end="-")
    
```

<h2 style="color:#C83F49"> Dictionaries </h2>

**Unordered:**
Unlike sequences such as lists and tuples, dictionaries do not maintain any specific order for their elements. This means that the order of key-value pairs may change over time.

**Mutable:**
Dictionaries can be modified after creation. You can add, update, or remove key-value pairs from a dictionary.

**Key-Value Pairs:**
Each element in a dictionary consists of a key and its associated value. Keys are used to access the corresponding values.

**Unique Keys:**
Keys within a dictionary must be unique. If you try to insert a duplicate key, it will overwrite the existing value associated with that key.

**Keys and Values Can Be of Any Data Type:**
Keys can be of any immutable data type like integers, strings, tuples, while values can be of any data type


<h3 style="color:#C1A1BD">1. Initialization</h3>

```python
# curly brackets

dict1 = {1:"Aaditya",
         2:"Aryan",
         3:"Satyam"}
print(dict1)

# dict() constructor

dict2 = dict([(1,"Aaditya"),(2,"Aryan"),(3,"Satyam")])
print(dict2)

dict3 = dict(key1='Aaditya', key2='Aryan', key3='Satyam')
print(dict3)

# zip method

key = [1,2,3]
value = ["Aaditya", "Aryan", "Satyam"]

dict4 = dict(zip(key,value))
print(dict4)

# dictionary comprehension

squares = {x: x*x for x in range(1, 6)}
print(squares)

dict5 = dict((key, value) for key, value in [(1,"Aaditya"),(2,"Aryan"),(3,"Satyam")])
print(dict5)

dict6 = {key: value for key, value in [(1,"Aaditya"),(2,"Aryan"),(3,"Satyam")]}
print(dict6)
```

<h3 style="color:#C1A1BD">2. Mapping with Dictionary</h3>

```python
# Original dictionary
my_dict = {'a': 1, 'b': 2, 'c': 3}
print(my_dict)

# Define a function to square the values
def square(x):
    return x ** 2

# Apply the function to each value using dictionary comprehension
mapped_dict = {key : square(value) for key, value in my_dict.items()}

print(mapped_dict) 
```

<h3 style="color:#C1A1BD">3. Lambda with Dictionary</h3>

```python
# Original dictionary
my_dict = {'a': 1, 'b': 2, 'c': 3}
print(my_dict)

# Apply the function to each value using map() and lambda function
mapped_dict = {key: value ** 2 for key, value in my_dict.items()}

print(mapped_dict)  
```

<h3 style="color:#C1A1BD">4. Accessing </h3>

```python
# square brackets

dict1 = {1:"Aaditya",
         2:"Aryan",
         3:"Satyam"}

print(f"printing the element with key 2 : {dict1[2]}")

dict2 = {('a', 'b'): 1, ('c', 'd'): 2}
print(f"printing the element with key ('c', 'd') : {dict2[('c', 'd')]}\n")

dict3 = {1:[1,2,3],
         2:[4,5,6],
         ('hello'):[7,8,9]}
print(f"printing the first element with key 1 : {dict3[1][0]}")
print(f"printing the second element with key 2 : {dict3[2][1]}")
print(f"printing the third element with key ('hello') : {dict3[('hello')][2]}\n")

# get()
# returns None if the key doesn't exist.

x = dict1.get(2)
print(f"{x}\n")

# using keys

for key in dict1.keys():
    value = print(dict1[key]) 
print("\n")

# using value

for value in dict1.values():
    print(value) 
print("\n")

# using key value pairs

for key, value in dict1.items():
    print(f"key {key} contains {value}")  
print("\n")

# pop()
# remove and return an element by key.

y = dict1.pop(3)
print(dict1)
print(y)
```

<h3 style="color:#C1A1BD">5. Removing Elements </h3>

```python
# del method
# You can remove a specific key-value pair from a dictionary

dict1 = {1:"Aaditya",
         2:"Aryan",
         3:"Satyam"}

del dict1[2]
print(dict1,"\n")

# pop()
# removes the specified key and returns the corresponding value
# If the key is not found, it raises a KeyError

dict2 = {1:"Aaditya",
         2:"Aryan",
         3:"Satyam"}

x = dict2.pop(2)
print(x)
print(dict2,"\n")

# popitem()
# removes and returns the last inserted key-value pair as a tuple. 

dict3 = {1:"Aaditya",
         2:"Aryan",
         3:"Satyam"}

y = dict3.popitem()
print(y)
print(dict3,"\n")

# clear()
# method removes all key-value pairs from the dictionary

dict3.clear()
print(dict3,"\n")

# using dictionary comprehension

my_dict = {'a': 1, 'b': 2, 'c': 3}
my_dict = {key: value for key, value in my_dict.items() if key != 'b'}  
print(my_dict)
```

<h3 style="color:#C1A1BD">5. Updating Elements </h3>

```python
dict1 = {1:"Aaditya",
         2:"Aryan",
         3:"Satyam"}

dict1[1] = "Aaditya Sinha"
dict1[2] = "Aryan Sharma"
dict1[3] = "Satyam Manhas"
print(dict1)
```

<h3 style="color:#C1A1BD">5. Dictionary Methods </h3>


```python
# key existence

dict1 = {'a':"Aaditya",
         'b':"Aryan",
         'c':"Satyam"}

if 'a' in dict1:
    print('Key "a" exists"\n')

# length

print(len(dict1),"\n")

# copying

dict2 = dict1.copy()
print(dict2)
```

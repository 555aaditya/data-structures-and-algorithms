<h2 style="color:#C83F49"> Exception Handling </h2>

```python
# it is used as a codeblock to handle any exceptions which may arise in a program so that the rest of the 
# program can run normally 
```

<h3 style="color:#C1A1BD">1. Try and Except</h3>

```python
# the codeblock which may return an error is entered within the try codeblock and except keyword handles the
# exceptions that may arise and returns any statement and the rest of the code is executed

x = input("Enter a number : ")

try:
    div = (100/int(x))
    print(div)
except Exception as i:
    print(i)

print("rest of the code")
    
```

<h3 style="color:#C1A1BD">2. Try, Except and Finally</h3>

```python
# with try, except and finally we add the codeblock which may give an error within the try codeblock and any excpetion 
# is handled by exception keyword however the finally codeblock is always executed if there is an error or not

def fun1():
    y = input("Enter a number : ")
    z = [1,2,3,4,5]
    try:
        print(z[int(y)])
        return 1
    except Exception as i:
        print(i)
        return 0
    finally:
        print("This always executes")
        
    print("rest of the code")

x = fun1()
print(x)

# here we can see the last print statement does not execute that is why we use a finally statement
# hence it returns 1 and the program ends
```

<h3 style="color:#C1A1BD">3. Custom Errors</h3>

```python
# using raise we can raise custom errors
x = int(input("Enter a number : "))

if (x<10 or x>50):
    raise ValueError("Value should be between 10 and 50")
```

<h2 style="color:#C83F49"> Map, Filter and Reduce </h2>

```python
# they are functions in Python that are used for functional programming and working with iterables. 
# They provide a concise and efficient way to perform operations on sequences like lists, tuples, and sets.
```

<h3 style="color:#C1A1BD">1. Map</h3>

```python
# map(function, iterable)
# applies the given function to each item of the iterable and returns an iterator that yields the results

numbers = [1, 2, 3, 4, 5]
print(numbers)
squared = map(lambda x: x*x, numbers)
print(list(squared)) 
```

<h3 style="color:#C1A1BD">2. Filter</h3>

```python
# filter(function, iterable)
# constructs an iterator from those elements of the iterable for which the function returns true

numbers = [1, 2, 3, 4, 5]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
```

<h3 style="color:#C1A1BD">3. Reduce</h3>

```python
from functools import reduce

# reduce(function, iterable, initializer )
# applies a rolling computation to sequential pairs of values in an iterable, ultimately producing a single result

numbers = [1, 2, 3, 4, 5]
sum_all = reduce(lambda x, y: x + y, numbers, 0)
print(sum_all)  
```

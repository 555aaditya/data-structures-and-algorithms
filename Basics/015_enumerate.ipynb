<h2 style="color:#C83F49"> Enumerate </h2>

```python
# enumerate() is used to iterate over a sequence while keeping track of the index of each element
# enumerate(sequence, start=0)

lst1 = ['Apple','Mango','Watermelon','Kiwi','Strawberry']
lst2 = ['Red','Blue','Green','Brown','Orange']

for i,j in enumerate(lst1):
    print(f'Index {i} : {j}')

print("")

for x,y in enumerate(lst2, start=1):
    print(f'Index {x} : {y}')
```

<h2 style="color:#C83F49"> Maths module in Python </h2>

<h3 style="color:#C1A1BD">1. Round</h3>

```python
import math

x = 123.189
y = 197.289

print(round(x,1))
print(round(y,2))
print(round(y,-1))
```

<h3 style="color:#C1A1BD">2. Ceiling</h3>

```python
import math

x = 2.8
y = 3.1

print(math.ceil(x))
print(math.ceil(y))
```

<h3 style="color:#C1A1BD">3. Floor</h3>

```python
import math

x = 2.8
y = 3.1

print(math.floor(x))
print(math.floor(y))
```

<h3 style="color:#C1A1BD">4. Absolute</h3>

```python
import math

x = -10
y = 5.9

print(abs(x))
print(abs(y))
```

<h3 style="color:#C1A1BD">5. Power</h3>

```python
import math

x = 3

print(pow(x,2))
```

<h3 style="color:#C1A1BD">6. Square Root</h3>

```python
import math

x = 25

print(math.sqrt(x))
```

<h3 style="color:#C1A1BD">7. Max</h3>

```python
import math

x = 1
y = 2
z = 3

print(max(x,y,z))
```

<h3 style="color:#C1A1BD">8. Min</h3>

```python
import math

x = 1
y = 2
z = 3

print(min(x,y,z))
```

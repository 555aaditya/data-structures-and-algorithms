 <h2 style="color:#C83F49"> Decorators </h2>

```python
# allows you to extend or modify the behavior of functions or methods without changing their implementation.
# Decorators are implemented as functions themselves and are used to wrap other functions or methods 
# to add some additional functionality.

def decorator_function(original_function):
    def wrapper_function(*args, **kwargs):
        # Additional functionality here
        return original_function(*args, **kwargs)
    return wrapper_function
```

```python
def greet(fx):
    def mfx(*args,**kwargs):
        print("Hello")
        fx(*args,**kwargs)
        print("Bye")
    return mfx


@greet
def sum(a,b):
    print(a+b)

sum(3,4)
```

```python
# https://www.youtube.com/watch?v=BE-L7xu8pO4

import time

# defining the decorator function
def timer(fx):
    def mfx(*args,**kwargs):
        t1 = time.time()
        fx(*args,**kwargs)
        t2 = time.time() - t1
        print(f"time taken {t2} seconds")
    return mfx


@timer
def sum(a,b):
    return a+b

sum(4,4)
```

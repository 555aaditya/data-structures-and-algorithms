<h2 style="color:#C83F49"> If Statements </h2>

<h3 style="color:#C1A1BD">1. If Statement</h3>

```python
# returns the code block if the condition is true

x = int(input("What is your age :"))

if (x >= 18):
    print("you are an adult")
```

<h3 style="color:#C1A1BD">2. If-Else Statement</h3>

```python
# else code block is returned when if condition returns false

x = int(input("What is your age :"))

if (x >= 18):
    print("you are an adult")
else:
    print("you are under age")
```

<h3 style="color:#C1A1BD">3. If-Elif-Else Statement</h3>

```python
# used to check for multiple conditions

x = float(input("enter your grade"))

if (x >= 9 and x <= 10):
    print("A")

elif (x >= 8 and x < 9):
    print("B")

elif (x >= 7 and x < 8):
    print("C")

elif (x >= 6 and x < 7):
    print("D")

elif (x >= 5 and x < 6):
    print("E")

elif (x < 5 and x >= 0):
    print("F")

else:
    print("Wrong Entry")
```

<h2 style="color:#C83F49"> Range function </h2>

<h3 style="color:#C1A1BD">1. One Arguement</h3>

```python
# single arguement gives the last element and it is exclusive

for i in range(5):
    print(i)
```

<h3 style="color:#C1A1BD">2. Two Arguements</h3>

```python
# two arguement gives the start and stop

for i in range(1,6):
    print(i)
```

<h3 style="color:#C1A1BD">3. Three Arguements</h3>

```python
# the three arguements give the start, stop and step size

for i in range(0,21,2):
    print(i)
```

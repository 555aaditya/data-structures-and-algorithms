<h2 style="color:#C83F49"> Strings and String Methods in Python </h2>

```python
x = "This is a string"
print(x)
print("Hello World")
```

<h3 style="color:#C1A1BD">1. Capitalize</h3>

```python
# used to capitalize the first character
x = "aaditya"
print("before capitalize :",x)
print("after capitalize :",x.capitalize())
```

<h3 style="color:#C1A1BD">2. Length</h3>

```python
# used to find the length of string
x = "apple"
print("length of the string is :",len(x))
```

<h3 style="color:#C1A1BD">3. Find</h3>

```python
# used to return the index of the character
x = "apple"
print("index of l :",x.find("l"))
```

<h3 style="color:#C1A1BD">4. Upper</h3>

```python
# used to capitalize all the characters
x = "apple"
print("upper-case of apple is :",x.upper())
```

<h3 style="color:#C1A1BD">5. Lower</h3>

```python
# used to lower all the characters
x = "APPLE"
print("lower-case of APPLE is :",x.lower())
```

<h3 style="color:#C1A1BD">6. isDigit</h3>

```python
# used to check if a string is an integer or not
x = "hello"
y = "1234"
print("is hello a digit :",x.isdigit())
print("is 1234 a digit :",y.isdigit())
```

<h3 style="color:#C1A1BD">7. isAlpha</h3>

```python
# checks if a string is an alphabet
x = "hello"
y = "1234"
print("is hello an alphabet :",x.isalpha())
print("is 1233 an alphabet ;",y.isalpha())
```

<h3 style="color:#C1A1BD">8. count</h3>

```python
# used to count a particular character in a string
x = "heellooo"
print("count of o in heellooo :",x.count("o"))
```

<h3 style="color:#C1A1BD">9. replace</h3>

```python
# used to replace a particular character in a string
x = "heellooo"
print("replacing l with j :",x.replace("l","j"))
```

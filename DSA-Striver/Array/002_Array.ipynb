<h1 style="color:#fc0000"> 1. Largest Element in an Array</h1>

<h3 style="color:#cccccc"> O(n) -> one iteration</h3>

```python
def largestElement(arr: list) -> int:
    largest = arr[0]

    for i in range(1,len(arr)):
        if arr[i] > largest:
            largest = arr[i]
    return largest

if __name__ == "__main__":
    arr = [1,2,4,2,7,5,3]
    largest = largestElement(arr)
    print(largest)
```

<h1 style="color:#fc0000"> 2. Second Largest Element in an Array</h1>

<h3 style="color:#cccccc"> O(2n) -> two iterations</h3>

```python
def secondLargestElement(arr: list) -> int:
    largest = arr[0]

    for i in range(1,len(arr)):
        if arr[i] > largest:
            largest = arr[i]

    secondLargest = float('-inf')
    for i in range(len(arr)):
        if(arr[i]>secondLargest and arr[i]!=largest):
            secondLargest = arr[i]
    return secondLargest

if __name__ == "__main__":
    arr = [1,2,7,2,7,5,3]
    secondLargest = secondLargestElement(arr)
    print(secondLargest)
```

<h3 style="color:#cccccc"> O(n) -> one iteration</h3>

```python
def secondLargestElement(arr: list) -> int:

    if(len(arr)) < 2: return None
        
    largest = arr[0]
    secondLargest = float('-inf')
        
    for i in range(len(arr)):
        if(arr[i]>largest):
            secondLargest = largest
            largest = arr[i]

        elif(arr[i]<largest and arr[i]>secondLargest):
            secondLargest = arr[i]
    return secondLargest if secondLargest != float('-inf') else None

if __name__ == "__main__":
    arr = [1,2,3,6,4,7,2,4]
    secondLargest = secondLargestElement(arr)
    print(secondLargest)
```

<h1 style="color:#fc0000"> 3. Check if Array is Sorted</h1>

<h3 style="color:#cccccc"> O(n) -> one iteration</h3>

```python
def checkSorted(arr: list) -> bool:
    for i in range(1,len(arr)):
        if(arr[i]>=arr[i-1]):
            pass
        else:
            return False
    return True

if __name__ == '__main__':
    arr = [1, 2, 2, 3, 4, 5, 5]
    check = checkSorted(arr)
    print(check)
```

<h1 style="color:#fc0000"> 4. Check if Array can be Sorted after rotation</h1>

<h3 style="color:#cccccc"> O(4n) -> four iterations</h3>

```python
# check if array is sorted

def checkSorted(arr: list) -> bool:
    for i in range(1,len(arr)):
        if(arr[i]>=arr[i-1]):
            pass
        else:
            return False
    return True

def ifPossible(arr: list) -> bool:

    if(checkSorted(arr) == True):
        return True # already sorted, roatation not needed

    # finding the breakpoint

    breakpoint = -1
    for i in range(len(arr)-1):
        if(arr[i]>arr[i+1]):
            breakpoint = i
            break

    # rotating the array

    rotated = arr[breakpoint+1:] + arr[:breakpoint+1]

    # checking if rotated array is sortable

    if(checkSorted(rotated) == True):
        return True
    else:
        return False

if __name__ == '__main__':
    arr = [5,6,1,2,3,4]
    check = ifPossible(arr)
    print(check)
```

<h1 style="color:#fc0000"> 5. Remove duplicates from a sorted array</h1>

<h3 style="color:#cccccc"> Brute Force Approach - copying unique elements to a new array</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def removeDuplicates(arr: list[int]) -> list[int]:
    newArr = []
    newArr.append(arr[0])
    size = 1

    for i in range(1,len(arr)):
        if(newArr[size-1] != arr[i]):
            newArr.append(arr[i])
            size+=1

    print(f"Array without duplicates is {newArr} of size {size}")

if __name__ == '__main__':
    arr = [1,2,2,3,4,5,5]
    newArr = removeDuplicates(arr)
```

<h3 style="color:#cccccc"> In-Place Approach - placing unique elements at the start</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def removeDuplicates(arr: list[int]) -> int:
    size = 1

    for i in range(1,len(arr)):
        if(arr[i] != arr[size-1]):
            arr[size] = arr[i]
            size+=1

    return size, arr

if __name__ == '__main__':
    arr = [1,2,2,2,3,3,4,5,5]
    newArr = removeDuplicates(arr)
    print(newArr)
```

<h1 style="color:#fc0000"> 6. Move zeroes to the end of an Array</h1>

```python
def moveZero(arr: list[int]) -> list[int]:
    # last non-zero index
    j = 0 
    
    for i in range(len(arr)):
        if(arr[i] != 0):
            arr[i], arr[j] = arr[j], arr[i]
            j+=1
    return arr

if __name__ == '__main__':
    arr = [1,2,0,1,4,5,0,6,0,8]
    newArr = moveZero(arr)
    print(newArr)
```

<h1 style="color:#fc0000"> 7. Left rotate an array by one</h1>

```python
def leftRotate(arr: list[int]) -> list[int]:
    # first element
    j = arr[0]

    for i in range(1,len(arr)):
        arr[i-1] = arr[i]

    arr[-1] = j
    return arr

if __name__ == '__main__':
    arr = [5,1,2,3,4]
    newArr = leftRotate(arr)
    print(newArr)
```

<h1 style="color:#fc0000"> 8. Left rotate an array by K</h1>

<h3 style="color:#cccccc"> Brute Force - calling left rotate by one K times</h3>

<h3 style="color:#cccccc"> Time Complexity - O(nk)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def leftRotate(arr: list[int]) -> list[int]:
    # first element
    j = arr[0]

    for i in range(1,len(arr)):
        arr[i-1] = arr[i]

    arr[-1] = j
    return arr

def leftRotateK(arr: list[int], k: int) -> list[int]:
    for i in range(k):
        leftRotate(arr)
    return arr

if __name__ == '__main__':
    arr = [5,1,2,3,4]
    newArr = leftRotateK(arr,3)
    print(newArr)
```

<h3 style="color:#cccccc"> Reverse - reverse array at k on both the side and reverse entire array</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def reverse(arr: list[int], low: int, high: int) -> None:
    while(low < high):
        arr[low], arr[high] = arr[high], arr[low]
        high-=1
        low+=1

def leftRotate(arr: list[int], k: int) -> None:
    reverse(arr,0,k-1)
    reverse(arr,k,len(arr)-1)
    reverse(arr,0,len(arr)-1)

if __name__ == '__main__':
    arr = [1,2,3,4,5,6,7,8,9]
    k = 4
    leftRotate(arr,k)
    print(arr)
```

<h1 style="color:#fc0000"> 9. Missing number in an array</h1>

<h3 style="color:#cccccc"> Brute Force Approach - Iterate through the array and check for the missing number</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def findMissing(arr: list[int]) -> int:
    n = len(arr)+1

    for i in range(len(arr)+1):
        if i not in arr:
            return i

if __name__ == '__main__':
    arr = [1,2,0,4]
    missing = findMissing(arr)
    print(missing)
```

<h3 style="color:#cccccc"> Hash Set Approach - Check for the missing value in a hash set</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def findMissing(arr: list[int]) -> int:
    n = len(arr)+1
    numSet = set(arr)
    
    for i in range(n):
        if i not in numSet:
            return i

if __name__ == '__main__':
    arr = [1,2,0,4]
    missing = findMissing(arr)
    print(missing)
```

<h3 style="color:#cccccc"> Optimised Approach - Check for missing value using sum formula</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>
<h3 style="color:#cccccc"> Missing Number = \(\frac{n \times (n+1)}{2} - \text{Actual Sum}\)</h3>

```python
def findMissing(arr: list[int]) -> int:
    n = len(arr)
    finalSum = (n*(n+1))/(2)
    actualSum = sum(arr)
    return int(finalSum - actualSum)

if __name__ == '__main__':
    arr = [1,2,0,4]
    missing = findMissing(arr)
    print(missing)
```

<h1 style="color:#fc0000"> 10. Maximum consecutive 1s in an array</h1>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def maxOne(arr: list[int]) -> int:
    counter = 0
    maximum = 0
    for i in range(len(arr)):
        if(arr[i] == 1):
            counter+=1
            maximum = max(counter,maximum)
        else:
            counter = 0
    return maximum

if __name__ == '__main__':
    arr = [1,0,1,1,1,0,0,1,1,0,1,1]
    maximum = maxOne(arr)
    print(maximum)
```

<h1 style="color:#fc0000"> 11. Find the number that appears once, and other numbers twice.</h1>

<h3 style="color:#cccccc"> Brute Force Approach - Iterate through the array and check for the missing number</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n&sup2)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def getSingleElement(arr):
    n = len(arr)
    
    for i in range(n):
        num = arr[i] 
        cnt = 0
        
        for j in range(n):
            if arr[j] == num:
                cnt += 1

        if cnt == 1:
            return num

    return -1

if __name__ == '__main__':
    arr = [4, 1, 2, 1, 2]
    ans = getSingleElement(arr)
    print(ans)
```

<h3 style="color:#cccccc"> Hash Map - Store value and the count of the value in a hash set

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(n)</h3>

```python
def getSingleElement(arr: list[int]) -> int:
    counter = {}

    for i in arr:
        counter[i] = counter.get(i,0) + 1

    for key, value in counter.items():
        if value == 1:
            return key

if __name__ == '__main__':
    arr = [1,2,1,2,3,5,4,3,5]
    singleElement = getSingleElement(arr)
    print(singleElement)
```

<h3 style="color:#cccccc"> Bit Manipulation using XOR</h3>
<h3 style="color:#cccccc">a ^ a = 0</h3> 
<h3 style="color:#cccccc">0 ^ b = b</h3>

<h3 style="color:#cccccc"> Time Complexity - O(n)</h3>
<h3 style="color:#cccccc"> Space Complexity - O(1)</h3>

```python
def getSingleElement(arr: list[int]) -> int:
    unique = 0
    for num in arr:
        unique ^= num

    return unique

if __name__ == '__main__':
    arr = [1,2,1,2,3,5,4,3,5]
    singleElement = getSingleElement(arr)
    print(singleElement)
```

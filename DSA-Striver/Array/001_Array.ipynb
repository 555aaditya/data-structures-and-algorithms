<h3 style="color:#42f566"> 1. Searching an element in an Array </h3>

```python
def find(a: list, b: int) -> int:
    for i in range(len(a)):
        if a[i] == b:
            return i
    return -1

if __name__ == "__main__":
    a = [1,2,4,3,6,7,5,9]
    b = 3
    if find(a,b) == -1:
        print("element not found")
    else:
        print(f"element found at index {find(a,b)}")
```

<h3 style="color:#42f566"> 2. Inserting an element at a index </h3>

```python
def insert(a: list, value: int, index:int) -> list:
    if index > len(a)+1:
        print("Index out of bounds")
        return a
    else:
        a.append(0)
        for i in range(len(a)-1,index-1,-1):
            a[i] = a[i-1]
        a[i] = value
        return a

if __name__ == "__main__":
    a = [1,2,3,5,6,7]
    value = 4
    index = 3
    print(insert(a,value,index))
```

<h3 style="color:#42f566"> 3. Deleting an element from a index </h3>

```python
def delete(a: list, index: int) -> list:
    if index<0 or index>len(a):
        print("Index out of Bounds")
        return a
    else:
        for i in range(index,len(a)-1):
            a[i] = a[i+1]
        a.pop()
        return a

if __name__ == "__main__":
    a = [1,2,3,5,6,7]
    index = 3
    print(delete(a,index))
```

<h3 style="color:#ff8c42">comparing all the elements with each other and returning the index where no larger element is found</h3>

<h3 style="color:#ff8c42">O(n<sup>2</sup>)</h3>

```python
def largest(a: list) -> int:
    for i in range(len(a)):
        flag = True
        for j in range(len(a)):
            if a[j] > a[i]:
                flag = False
        if flag == True:
            return i

if __name__ == "__main__":
    a = [10, 30, 20, 50, 70, 90, 25, 33]
    print(f"The largest element is present at {largest(a)} and has a value of {a[largest(a)]}")
```

<h3 style="color:#42f566"> 6. Second Largest element in an array </h3>

<h3 style="color:#ff8c42">comparing all elements and ignoring the largest element</h3>

<h3 style="color:#ff8c42">O(n)</h3>

```python
def largest(a: list) -> int:
    index = 0
    for i in range(len(a)):  
        if a[i] > a[index]:
            index = i
    return index

def secondLargest(a: list) -> int:
    index = largest(a)
    index2 = None
    for i in range(len(a)):
        if (a[i] != a[index]):
            if a[i] > a[index2] or index2 == None:
                index2 = i
    return index2

if __name__ == "__main__":
    a = [1,1,1,1,1,1,1]
    if secondLargest(a) is not None:
        print(f"The second largest element is present at {secondLargest(a)} and has a value of {a[secondLargest(a)]}")
    else:
        print("The array does not have a second distinct largest element.")
```

```python
def secondLargest(a: list) -> int:
    largest = secondLargest = float('-inf')
    
    for i in a:
        if i > largest:
            secondLargest = largest
            largest = i
        elif i > secondLargest and i != largest:  
            secondLargest = i
    
    if secondLargest == float('-inf') or largest == secondLargest:
        return None 
        
    return secondLargest

if __name__ == "__main__":
    a = [1, 1, 1, 1, 1, 1]
    result = secondLargest(a)
    if result is not None:
        print(f"The second largest element is present and has a value of {result}")
    else:
        print("The array does not have a second distinct largest element.")
```

<h3 style="color:#42f566"> 7. Check if an array is sorted </h3>

<h3 style="color:#ff8c42">comparing all the elements with the next element and returning False if we encounter a smaller element</h3>

<h3 style="color:#ff8c42">O(n)</h3>

```python
def checkSort(arr: list) -> bool:
    for i in range(len(arr)-1):
        if arr[i] >= arr[i+1]:
            return False
    return True

if __name__ == "__main__":
    a = [10,30,20,40]
    print(checkSort(a))
```

<h3 style="color:#42f566"> 7. Reverse an Array </h3>

<h3 style="color:#ff8c42"> we run a loop for half the length of the array as long as index of high is greater and swap elements</h3>

<h3 style="color:#ff8c42">O(n)</h3>

```python
def reverse(arr: list) -> list:
    low = 0
    high = len(arr)-1
    while(high>low):
        arr[low], arr[high] = arr[high], arr[low]
        low+=1
        high-=1
    return arr

if __name__ == "__main__":
    a = [10,20,30,40]
    print(reverse(a))
```

<h3 style="color:#42f566"> 7. Remove duplicates from sorted array </h3>

<h3 style="color:#ff8c42"> compare the current element with the last element in temp for unique element append it to temp.</h3>

<h3 style="color:#ff8c42"> O(n)</h3>

```python
def removeDuplicate(a: list) -> list:
    
    temp = [a[0]] 
    for i in range(1, len(a)):
        if a[i] != temp[-1]:  
            temp.append(a[i])
    return temp

if __name__ == "__main__":
    a = [10,10,20,30,40,40,50]
    print(removeDuplicate(a))
```

<h3 style="color:#42f566"> 7. Move all zeroes to the end of an array </h3>

<h3 style="color:#ff8c42"> Iterate through the array and when it finds 0 and check all elements after and replace next non-zero with 0</h3>

<h3 style="color:#ff8c42"> O(n<sup>2</sup>)</h3>

```python
def moveZero(a: list[int]) -> list[int]:
    for i in range(len(a)):
        if (a[i] == 0):
            for j in range(i+1,len(a)):
                if(a[j] != 0):
                    a[i],a[j] = a[j],a[i]
    return a

if __name__ == "__main__":
    a = [10,10,0,0,40,0,50]
    print(moveZero(a))
```

<h3 style="color:#ff8c42"> Iterate through the array and when it finds a non-zero value it saves it's index and when it finds 0 it replaces it with the non-zero element</h3>

<h3 style="color:#ff8c42"> O(n)</h3>

```python
def moveZero(a: list[int]) -> list[int]:
    nonZeroindex = 0
    for i in range(len(a)):
        if a[i] != 0:
            a[nonZeroindex],a[i] = a[i],a[nonZeroindex]
            nonZeroindex += 1
    return a

if __name__ == "__main__":
    a = [10,10,0,0,40,0,50]
    print(moveZero(a))
```

<h3 style="color:#42f566"> 8. Left rotate an array by one</h3>

<h3 style="color:#ff8c42"> we shift the first element counter clockwise, hence it moves to the end and the last element becomes first  </h3>

<h3 style="color:#ff8c42"> O(1)  </h3>

```python
def leftRotate(a: list[int]) -> list[int]:
    a[-1],a[0] = a[0],a[-1]
    return a

if __name__ == "__main__":
    a = [10,20,40,50]
    print(leftRotate(a))
```

<h3 style="color:#ff8c42"> we store the first element, and shift all subsequent element one index back and replace last index with stored first element   </h3>

<h3 style="color:#ff8c42"> O(n)  </h3>

```python
def leftRotate(arr: list[int]) -> list[int]:
    temp = arr[0]
    for i in range(1,len(arr)):
        arr[i-1] = arr[i]
    arr[-1] = temp;
    return arr

if __name__ == "__main__":
    a = [10,20,40,50]
    print(leftRotate(a))
```

<h3 style="color:#42f566"> 9. Left rotate an array by k</h3>

<h3 style="color:#ff8c42">we move each element counter clockwise for k times until the elements are shifted k times</h3>

<h3 style="color:#ff8c42"> O(kn)  </h3>

```python
def leftRotate(arr: list[int], k: int) -> list[int]:
    while (k!=0):
        temp = arr[0]
        for i in range(1,len(arr)):
            arr[i-1] = arr[i]
        arr[-1] = temp
        k-=1
    return arr

if __name__ == "__main__":
    a = [10,20,40,50]
    print(leftRotate(a,2))
```

<h3 style="color:#ff8c42"> arr[k:] creates a new list containing elements from index k to the end of the list and arr[:k] creates a new list containing elements from the start of the list to index k-1 and we concatenate them together.
 </h3>

<h3 style="color:#ff8c42"> O(n)  </h3>

```python
def leftRotate(arr: list[int], k: int) -> list[int]:
    k = k%len(arr)
    return arr[k:] + arr[:k]

if __name__ == "__main__":
    a = [10,20,40,50,60]
    print(leftRotate(a,2))
```

<h3 style="color:#42f566"> 10. Leader in an array</h3>

<h3 style="color:#ff8c42">for each element in the array we check if any element is larger to the right and break if found, and if no larger element is to the right we return the element</h3>

<h3 style="color:#ff8c42"> O(n<sup>2</sup>)</h3>

```python
def leader(arr: list[int]):
    for i in range(len(arr)):
        flag = False
        for j in range(i+1,len(arr)):
            if (arr[j] >= arr[i]):
                flag = True
                break
        if (flag == False):
            print(arr[i])

if __name__ == "__main__":
    a = [10,20,80,50,60]
    leader(a)
```

```python
def leader(arr: list[int]):
    print(arr[-1])
    max = arr[-1]
    for i in range(len(arr)-2,0,-1):
        if arr[i] > max:
            print(arr[i])
            max = arr[i]

if __name__ == "__main__":
    a = [10,20,80,50,60]
    leader(a)
```

<h3 style="color:#42f566"> 11. Maximum difference in an array</h3>

<h3 style="color:#ff8c42">we compare each element with every element towards the right and store the maximum differences</h3>

<h3 style="color:#ff8c42">O(n<sup>2</sup>)</h3>

```python
def maxDiff(arr: list[int]) -> int:
    max = float('-inf')
    for i in range(len(arr)):
        for j in range(i+1,len(arr)):
            if (arr[j]-arr[i]) >= max:
                max = (arr[j]-arr[i])
    return max

if __name__ == "__main__":
    a = [80,70,60,40,10,8]
    print(maxDiff(a))
```

<h3 style="color:#ff8c42">we find the difference between the current element and the minimum element and change the maximum difference and minimum value to get the answer</h3>

<h3 style="color:#ff8c42">O(n)</h3>

```python
def maxDiff(arr: list[int]) -> int:
    min_value = arr[0]
    max_diff = float('-inf')
    for i in range(1,len(arr)):
        current_diff = arr[i] - min_value
        if (current_diff > max_diff):
            max_diff = current_diff
        if (arr[i] < min_value):
            min_value = arr[i]
    return max_diff

if __name__ == "__main__":
    a = [90, 40, 20, 70, 30]
    print(maxDiff(a))
```

<h3 style="color:#42f566"> 12. Frequency in a sorted array</h3>

<h3 style="color:#ff8c42">we check if the next element is equal to the previous and increment the frequency and the reset the frequency</h3>

<h3 style="color:#ff8c42">O(n)</h3>

```python
def frequency(arr: list[int]):
    frequency = 1
    i = 1
    while i < len(arr)+1:
        while i < len(arr) and arr[i] == arr[i - 1]:
            frequency += 1
            i += 1
        print(arr[i - 1], frequency)
        frequency = 1
        i += 1
    if (len(arr)==1):
        print(arr[-1], frequency)
    if not arr:  
        print("empty list")

if __name__ == "__main__":
    a = [10,10,20,20,20,30]
    frequency(a)
```

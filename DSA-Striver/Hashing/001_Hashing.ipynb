<h1 style="color:#fc0000"> 1. Counting the number of times an element is present in an array </h1>

```python
#size of array
n = int(input())

#array
arr = list(map(int,input().split()))

hash_map = {}

for num in arr:
    hash_map[num] = hash_map.get(num,0) + 1

#query
q = int(input())

for _ in range(q):
    number = int(input())
    print(hash_map.get(number,0))
```

<h1 style="color:#fc0000"> 2. Counting the number of times a character is present in a string </h1>

```python
#input string
string = input()

hash_map = {}

for char in string:
    hash_map[char] = hash_map.get(char,0)+1

# query
q = int(input())

for _ in range(q):
    char = input()
    print(f"{char} is present {hash_map.get(char,0)} times")
```

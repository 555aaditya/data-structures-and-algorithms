<h2 style="color:#1f73e0"> Creating Data Frames in Pandas </h2>

```python
import pandas as pd
import numpy as np
```

```python
# Describing Data

data1 = np.array([["Aaditya",19,6.4,"22BCB-1","Y"],["Aryan",20,4.5,"22BCB-1","Y"],["Satyam",21,0.33,"22BCB-1","N"]])
print(data1)
```

```python
# Building DataFrame

df = pd.DataFrame(data=data1, index=['S1','S2','S3'], columns=["Name","Age","CGPA","Section","TPP"])
print(df)
```

```python
# Indexing

# using multiple column names
print(df[["Name","CGPA","TPP"]],"\n")

# using single column name
print(df["Name"],"\n")

# using loc for single row with row-index
print(df.loc['S2'],"\n")

# using loc for multiple rows with row-index
print(df.loc[['S2','S3']],"\n")

# using iloc for a set of data values
# [row_start:row_end,column_start:column_end]
print(df.iloc[1:2,0:2]) 
```

```python
# Converting DataFrame to array

print(df.iloc[:,0:3],"\n")
print(df.iloc[:,0:3].values)
```

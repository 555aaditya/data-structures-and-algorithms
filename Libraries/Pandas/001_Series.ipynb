<h2 style="color:#1f73e0"> Creating Series in Pandas </h2>

```python
import pandas as pd
import numpy as np
```

```python
# Describing Data

data1 = np.array(["IND","AUS","NZ","SA","PAK","BAN"])

# Creating a series

df = pd.Series(data1,index=["a","b","c","d","e","f"])
print(df,"\n")

print(type(df),"\n")
print(len(df))
```

<h2 style="color:#1f73e0"> Series Methods </h2>

```python
# accessing elements

# using index value
print(df['a']) 

# using iloc
print(df.iloc[3])

# using loc
print(df.loc['b'],"\n")

# using head and tail
print(df.head(4),"\n")
print(df.tail(3))
```

```python
# returning array from Series
print(df.values)

# returning index from Series
print(df.index)

# returning datatype of Series
print(df.dtype)
```

```python
# Information about the Series

# describe
print(df.describe(),"\n")

# shape
print(df.shape,"\n")

# size
print(df.size)
```

```python
# applying a function to all the elements
# it iterates through each element of the Series

data2 = np.array([1,2,3,4,5])
df1 = pd.Series(data2)
print(df1)

# adding 1 to all the elements

def add1(x):
    return x+1

df2 = df1.apply(add1)
print(df2)
```
